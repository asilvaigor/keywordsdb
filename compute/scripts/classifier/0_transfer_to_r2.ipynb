# 0. Copy dataset to Cloudflare R2


```python
# uv pip install awscli

import json
import os
import subprocess

with open("../../../tf/cloudflare-local-account-key.tfstate") as f:
    creds = json.load(f)

env = {**os.environ, "AWS_ACCESS_KEY_ID": creds["access_key_id"], "AWS_SECRET_ACCESS_KEY": creds["secret_access_key"]}

subprocess.run(
    [
        "aws",
        "s3",
        "sync",
        "../../../data/2026-03-30/",
        "s3://keywordsdb/keywordsdb/2026-03-30/",
        "--endpoint-url",
        creds["endpoint_url"],
        "--region",
        "auto",
    ],
    env=env,
    check=True,
)
```
