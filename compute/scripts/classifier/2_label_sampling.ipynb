# 2. Sampling

Collects a sample of keywords to be labeled.

To be run locally anywhere.


```python
import polars as pl

df: pl.DataFrame = (
    pl.scan_parquet("../../../data/2026-03-30/*.parquet")
    .filter(pl.col("volume_estimated_google") > 10000)
    .select("keyword", "volume_estimated_google")
    .sort("volume_estimated_google", descending=True)
    .collect()
)  # ty:ignore[invalid-assignment]
```

```python
df.with_columns(n_chars=pl.col("keyword").str.len_chars()).select("volume_estimated_google", "n_chars").describe()
```

```python
df.sample(100000, seed=42).write_parquet("../../../data/sample.parquet")
```
